```python
import random

historico = []

while True:
    numero = random.randint(0, 15)
    tentativas = 0

    print("\nTente adivinhar o número de 0 a 15!")

    while True:
        palpite = int(input("Digite seu palpite: "))
        tentativas += 1

        if palpite == numero:
            print("Você acertou!")
            print(f"Tentativas: {tentativas}")

            historico.append((numero, tentativas))

            with open("historico.txt", "a") as arquivo:
                arquivo.write(f"Número: {numero} - Tentativas: {tentativas}\n")

            break

        elif palpite < numero:
            print("O número é maior!")

        else:
            print("O número é menor!")

    continuar = input("Deseja jogar novamente? Brendha: ")

    if continuar != "s":
        break

print("\nHISTÓRICO DOS ACERTOS:")

for numero, tentativas in historico:
    print(f"Número: {numero} - Tentativas: {tentativas}")
```

```python
import csv

compras = []

while True:
    item = input("Digite o nome do item (ou 'fim' para terminar): ")

    if item == "fim":
        break

    quantidade = int(input("Digite a quantidade: "))
    valor = float(input("Digite o valor do item: "))

    total = quantidade * valor

    compras.append([item, quantidade, valor, total])

valor_total = 0

for compra in compras:
    valor_total += compra[3]

print("\nLISTA DE COMPRAS:")

for compra in compras:
    print(f"Item: {compra[0]} | Quantidade: {compra[1]} | Valor: R$ {compra[2]:.2f} | Total: R$ {compra[3]:.2f}")

print(f"\nValor total das compras: R$ {valor_total:.2f}")

with open("lista_compras.csv", "w", newline="") as arquivo:
    escritor = csv.writer(arquivo)

    escritor.writerow(["Item", "Quantidade", "Valor", "Total"])

    for compra in compras:
        escritor.writerow(compra)

print("\nLista salva no arquivo lista_compras.csv!")
```
