```python
# Usamos o comando return para retornar valores de uma função e encerrer sua execução
# Função para calcular percentual de um valor qualquer
# É possível deixar um valor fixo para um argumento,
# quando ele não é informado na chamada, o valor padrão é utilizado
def calcula_percento(valor, percentual=100):
  quanto = valor * percentual / 100
  return quanto

print ('20% de 300 são: {:10.2f}'.format (calcula_percento (300,20)))

print ('100% de 50 são: {:10.2f}'.format (calcula_percento (50)))
```

```python
quantidade = int(input("Digite a quantidade de números primos: "))

encontrados = 0
numero = 2

while encontrados < quantidade:
    primo = True

    if numero == 0 or numero == 1:
        primo = False
    elif numero > 2 and numero % 2 == 0:
        primo = False
    else:
        for divisor in range(3, numero, 2):
            if numero % divisor == 0:
                primo = False
                break

    if primo:
        print(numero)
        encontrados += 1

    numero += 1
```

```python
import math

n = float(input("Digite um número: "))

# Valor inicial para b
b = n

# Calcula o primeiro valor de p
p = (b + (n / b)) / 2

# Repete o cálculo até atingir a precisão desejada
while abs(n - p ** 2) >= 0.0001:
    b = p
    p = (b + (n / b)) / 2

# Resultado usando o método de Newton
print("Raiz quadrada pelo método de Newton:", p)

# Resultado usando a função sqrt()
print("Raiz quadrada usando sqrt:", math.sqrt(n))
```
