```python
# Função para calcular a maior nota
def calcular_maior_nota(notas):
    return max(notas)


# Função para calcular a menor nota
def calcular_menor_nota(notas):
    return min(notas)


# Função para calcular a média da turma
def calcular_media(notas):
    return sum(notas) / len(notas)


# Função para calcular a quantidade de estudantes aprovados
def calcular_aprovados(notas):
    aprovados = 0

    for nota in notas:
        if nota >= 6.0:
            aprovados += 1

    return aprovados


# Função para calcular a quantidade de estudantes reprovados
def calcular_reprovados(notas):
    reprovados = 0

    for nota in notas:
        if nota < 6.0:
            reprovados += 1

    return reprovados


# Programa principal
quantidade = int(input("Digite a quantidade de estudantes: "))

notas = []

for i in range(quantidade):
    nota = float(input(f"Digite a nota do estudante {i + 1}: "))
    notas.append(nota)


# Cálculos
maior_nota = calcular_maior_nota(notas)
menor_nota = calcular_menor_nota(notas)
media = calcular_media(notas)
aprovados = calcular_aprovados(notas)
reprovados = calcular_reprovados(notas)


# Exibindo os resultados
print("\n--- RESULTADO DA TURMA ---")
print(f"Notas: {notas}")
print(f"Maior nota: {maior_nota:.2f}")
print(f"Menor nota: {menor_nota:.2f}")
print(f"Média da turma: {media:.2f}")
print(f"Quantidade de estudantes aprovados: {aprovados}")
print(f"Quantidade de estudantes reprovados: {reprovados}")
```

```python
# Função para cadastrar produtos
def cadastrar_produtos():
    produtos = {}

    quantidade = int(input("Digite a quantidade de produtos: "))

    for i in range(quantidade):
        print(f"\nProduto {i + 1}")

        codigo = input("Digite o código do produto: ")
        nome = input("Digite o nome do produto: ")
        preco = float(input("Digite o preço do produto: "))
        estoque = int(input("Digite a quantidade em estoque: "))

        produtos[codigo] = {
            "nome": nome,
            "preco": preco,
            "estoque": estoque
        }

    return produtos


# Função para consultar um produto pelo código
def consultar_produto(produtos):
    codigo = input("\nDigite o código do produto que deseja consultar: ")

    if codigo in produtos:
        print("\n--- PRODUTO ENCONTRADO ---")
        print(f"Código: {codigo}")
        print(f"Nome: {produtos[codigo]['nome']}")
        print(f"Preço: R$ {produtos[codigo]['preco']:.2f}")
        print(f"Quantidade em estoque: {produtos[codigo]['estoque']}")
    else:
        print("Produto não encontrado.")


# Função para calcular o valor total do estoque
def calcular_valor_total(produtos):
    total = 0

    for codigo in produtos:
        preco = produtos[codigo]["preco"]
        estoque = produtos[codigo]["estoque"]

        total += preco * estoque

    return total


# Função para encontrar o produto de maior valor unitário
def produto_maior_valor(produtos):
    maior_codigo = None
    maior_preco = 0

    for codigo in produtos:
        preco = produtos[codigo]["preco"]

        if preco > maior_preco:
            maior_preco = preco
            maior_codigo = codigo

    return maior_codigo


# Programa principal

produtos = cadastrar_produtos()

consultar_produto(produtos)

valor_total = calcular_valor_total(produtos)

codigo_maior_valor = produto_maior_valor(produtos)

print("\n--- RESULTADOS ---")
print(f"Valor total do estoque: R$ {valor_total:.2f}")

print("\nProduto de maior valor unitário:")
print(f"Código: {codigo_maior_valor}")
print(f"Nome: {produtos[codigo_maior_valor]['nome']}")
print(f"Preço: R$ {produtos[codigo_maior_valor]['preco']:.2f}")
```
