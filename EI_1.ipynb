```python
# Função para calcular a maior nota
def calcular_maior_nota(notas):
    return max(notas)


# Função para calcular a menor nota
def calcular_menor_nota(notas):
    return min(notas)


# Função para calcular a média da turma
def calcular_media(notas):
    return sum(notas) / len(notas)


# Função para calcular a quantidade de estudantes aprovados
def calcular_aprovados(notas):
    aprovados = 0

    for nota in notas:
        if nota >= 6.0:
            aprovados += 1

    return aprovados


# Função para calcular a quantidade de estudantes reprovados
def calcular_reprovados(notas):
    reprovados = 0

    for nota in notas:
        if nota < 6.0:
            reprovados += 1

    return reprovados


# Programa principal
quantidade = int(input("Digite a quantidade de estudantes: "))

notas = []

for i in range(quantidade):
    nota = float(input(f"Digite a nota do estudante {i + 1}: "))
    notas.append(nota)


# Cálculos
maior_nota = calcular_maior_nota(notas)
menor_nota = calcular_menor_nota(notas)
media = calcular_media(notas)
aprovados = calcular_aprovados(notas)
reprovados = calcular_reprovados(notas)


# Exibindo os resultados
print("\n--- RESULTADO DA TURMA ---")
print(f"Notas: {notas}")
print(f"Maior nota: {maior_nota:.2f}")
print(f"Menor nota: {menor_nota:.2f}")
print(f"Média da turma: {media:.2f}")
print(f"Quantidade de estudantes aprovados: {aprovados}")
print(f"Quantidade de estudantes reprovados: {reprovados}")
```
