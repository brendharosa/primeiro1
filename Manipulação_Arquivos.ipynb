```python
import random
import json
import csv
```

```python
numeros = random.sample(range(1, 101), 20)

print("Lista de números aleatórios:")
print(numeros)
```

```python
dados = []

for numero in numeros:
    dados.append({
        "valor_original": numero,
        "valor_vezes_2": numero * 2,
        "valor_vezes_10": numero * 10
    })

for item in dados:
    print(item)
```

```python
with open("numeros.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write("Valor Original | Valor * 2 | Valor * 10\n")

    for item in dados:
        arquivo.write(
            f"{item['valor_original']} | "
            f"{item['valor_vezes_2']} | "
            f"{item['valor_vezes_10']}\n"
        )

print("Arquivo TXT salvo com sucesso!")
```

```python
with open("numeros.json", "w", encoding="utf-8") as arquivo:
    json.dump(dados, arquivo, indent=4, ensure_ascii=False)

print("Arquivo JSON salvo com sucesso!")
```

```python
with open("numeros.csv", "w", newline="", encoding="utf-8") as arquivo:
    campos = ["valor_original", "valor_vezes_2", "valor_vezes_10"]

    escritor = csv.DictWriter(arquivo, fieldnames=campos)

    escritor.writeheader()
    escritor.writerows(dados)

print("Arquivo CSV salvo com sucesso!")
```
