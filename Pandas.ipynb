# Criando um dataset importando um arquivo csv (não altere).

O arquivo *clima_historico_municipal.csv* contém um registro por município, identificado por código IBGE, nome e UF. Para cada município, a base informa o período de referência, a fonte climática, o método de processamento e um conjunto de variáveis mensais agregadas historicamente. As variáveis climáticas presentes são:

* tmin_*: temperatura mínima média mensal histórica
* tmax_*: temperatura máxima média mensal histórica
* temp_media_*: temperatura média mensal histórica, derivada de * tmin e tmax
* ppt_*: precipitação mensal histórica
* pet_*: evapotranspiração potencial mensal histórica
* aet_*: evapotranspiração real mensal histórica
* def_*: déficit hídrico mensal histórico

Cada sufixo mensal corresponde a um mês do ano:

*jan, fev, mar, abr, mai, jun, jul, ago, set, out, nov, dez*

Assim, por exemplo:

* temp_media_jan representa a média histórica de temperatura média para todos os janeiros do período processado
* ppt_mar representa a média histórica de precipitação para todos os meses de março do período processado
* def_set representa a média histórica de déficit hídrico para todos os meses de setembro do período processado

As unidades utilizadas são:
* temperaturas (*tmin, tmax, temp_media*): **graus Celsius (°C)**
* precipitação (*ppt*): **milímetros por mês (mm/mês)**
* evapotranspiração potencial (*pet*): **milímetros por mês (mm/mês)**
* evapotranspiração real (*aet*): **milímetros por mês (mm/mês)**
* déficit hídrico (*def*): **milímetros por mês (mm/mês)**

# Como renomear o nome das colunas

```python
'''
O comando rename permite alterar o nome de uma ou mais colunas do dataframe.
Neste exemplo, a coluna nome_municipio será alterada para municipio.
Info: Brendha
'''

# cria uma cópia do dataframe
df_copia = df.copy()

# renomeia a coluna
df_copia.rename(columns={'nome_municipio': 'municipio'}, inplace=True)

# visualiza o dataframe
df_copia.head()
```

# Como adicionar linhas ao dataframe

```python
'''
Para adicionar uma nova linha ao dataframe pode utilizar o comando loc.
A nova linha é inserida na última posição do dataframe.
Info: Brendha
'''

# cria uma cópia do dataframe
df_copia = df.copy()

# copia uma linha existente
nova_linha = df_copia.iloc[0].copy()

# altera algumas informações
nova_linha['codigo_ibge'] = 9999999
nova_linha['nome_municipio'] = 'Municipio Exemplo'

# adiciona a nova linha
df_copia.loc[len(df_copia)] = nova_linha

# visualiza as últimas linhas
df_copia.tail()
```

# Como adicionar colunas ao dataframe

```python
'''
Para adicionar uma nova coluna precisa informar o nome da nova coluna entre colchetes.
Neste exemplo será criada uma coluna com a temperatura média anual dos municípios.
Info: Brendha
'''

# cria uma cópia do dataframe
df_copia = df.copy()

# cria uma nova coluna
df_copia['temp_media_anual'] = df_copia[
    ['temp_media_jan', 'temp_media_fev', 'temp_media_mar',
     'temp_media_abr', 'temp_media_mai', 'temp_media_jun',
     'temp_media_jul', 'temp_media_ago', 'temp_media_set',
     'temp_media_out', 'temp_media_nov', 'temp_media_dez']
].mean(axis=1)

# visualiza o dataframe
df_copia.head()
```

# Como deletar linhas

```python
'''
O comando drop permite excluir linhas do dataframe.
Neste exemplo será removida a linha de índice 0.
Info: Brendha
'''

# cria uma cópia do dataframe
df_copia = df.copy()

# exclui a linha de índice 0
df_copia.drop(index=0, inplace=True)

# visualiza o dataframe
df_copia.head()
```

# Como deletar colunas

```python
'''
O comando drop pode ser utilizado para excluir colunas do dataframe.
Para isso é utilizado o parâmetro columns informando a coluna que será removida.
Info: Brendha
'''

# cria uma cópia do dataframe
df_copia = df.copy()

# exclui uma coluna
df_copia.drop(columns=['metodo_processamento'], inplace=True)

# visualiza o dataframe
df_copia.head()
```

# Como mesclar dois dataframe

```python
'''
O comando merge permite unir dois dataframes utilizando uma coluna em comum.
Neste exemplo os dataframes são unidos através da coluna codigo_ibge.
Info: Brendha
'''

# cria uma cópia do dataframe
df_copia = df.copy()

# cria dois dataframes menores
df_temperatura = df_copia[['codigo_ibge', 'nome_municipio', 'temp_media_jan']]
df_chuva = df_copia[['codigo_ibge', 'ppt_jan']]

# mescla os dois dataframes
df_mesclado = pd.merge(df_temperatura, df_chuva, on='codigo_ibge')

# visualiza o resultado
df_mesclado.head()
```
