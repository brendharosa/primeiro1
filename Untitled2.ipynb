```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dados = {
    "Mes":["Jan", "Fev", "Mar", "Abr", "Mai", "Jun"],
    "Notebook":[20,25,22,30,35,40],
    "Mouse": [20,55,48,60,65,70],
    "Teclado":[30,28,35,40,38,45]
}

df = pd.DataFrame(dados)
print(df)
```

```python
# Exibir as primeiras linhas
print("Primeiras linhas")
print(df.head())

# Exibir as dimenções do DataFrame
print("\nDimenções:")
print(df.shape)

# Exibir as estatísticas descritivas
print("\nEstatísticas descritivas:")
print(df.describe())
```

```python
# Indicadores de vendas com NumPy

produtos = ["Notebook", "Mouse", "Teclado"]

for produto in produtos:
    vendas = df [produto].to_numpy()

    print(f"\n{produto}")
    print(f"Média: {np.mean(vendas):.2f}")
    print(f"Maior venda: {np.max(vendas)}")
    print(f"Menor venda: {np.min(vendas)}")
```

```python
plt.plot(df["Mes"], df["Notebook"], marker="o", label="Notebook")
plt.plot(df["Mes"], df["Mouse"], marker="o", label="Mouse")
plt.plot(df["Mes"], df["Teclado"], marker="o", label="Teclado")

plt.title("Evolução das Vendas por Produto")
plt.xlabel("Meses")
plt.ylabel("Quantidade Vendida (unidades)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
totais = df[["Notebook", "Mouse", "Teclado"]].sum()

plt.bar(totais.index, totais.values)

plt.title("Total de Vendas por Produto")
plt.xlabel("Produtos")
plt.ylabel("Quntidade Total Vendida (unidades)")
plt.show()
```

```python
plt.hist(df["Notebook"], bins=5, edgecolor="black")

plt.title("Distribuição das Vendas Mensais de Notebooks")
plt.xlabel("Quantidade Vendida (unidades)")
plt.ylabel("Frequência")
plt.show()
```

```python
# Total de vendas por produto
totais = df[["Notebook", "Mouse", "Teclado"]].sum()

print("Totais vendido por produto:")
print(totais)

# Produto com maior volume total
print("\nProduto com maior volume total:")
print(totais.idxmax())

# Mês de maior venda de cada produto
print("\nMês de maior venda de cada produto:")

for produto in ["Notebook", "Mouse", "Teclado"]:
  indice = df[produto].idxmax()
  mes = df.loc[indice, "Mes"]
  venda = df.loc[indice, produto]

  print(f"{produto}: {mes}, com {venda} unidades")
```

```python

# Calcular a média mensal de notebooks
media = np.mean(df["Notebook"])

# Selecionar os meses abaixo da média
abaixo_media = df[df["Notebook"] < media]

# Criar o gráfico
plt.figure(figsize=(8, 5))

plt.bar(
    abaixo_media["Mes"],
    abaixo_media["Notebook"],
    label="Vendas abaixo da média"
)

plt.axhline(
    y=media,
    color="red",
    linestyle="--",
    label=f"Média mensal: {media:.2f}"
)

plt.title("Vendas de Notebooks Abaixo da Média")
plt.xlabel("Meses")
plt.ylabel("Quantidade Vendida (unidades)")
plt.ylim(0, 45)
plt.legend()
plt.grid(axis="y", linestyle=":", alpha=0.5)
plt.tight_layout()
plt.show()

# Exibir os resultados
print(f"Média mensal: {media:.2f} unidades")
print("\nMeses abaixo da média:")
print(abaixo_media[["Mes", "Notebook"]].to_string(index=False))



```

**Conclusão**

O mouse apresentou o maior volume de vendas, com 348 unidades, e junho foi o mês de melhor resultado para os três produtos. Os gráficos facilitaram a comparação das vendas e a identificação dos meses abaixo da média de notebooks. Assim, foi possível utilizar pandas, NumPy e Matplotlib para analisar e visualizar os dados.
